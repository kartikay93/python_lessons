## Encapsulation

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

person = Person("Kartik", 21)
print(person.name)
```

```python
dir(person)
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name # public variable
        self.age = age # public variable

def get_name(a):
    return a.name

person = Person("Kartik", 21)
get_name(person)
```

```python
class Person:
    def __init__(self, name, age, gender):
        self.__name = name # private variable
        self.__age = age # private variable
        self.gender = gender # public variable


person = Person("Kartik", 21, "Male")
```

```python
dir(person)
```

```python
class Person:
    def __init__(self, name, age, gender):
        self.__name = name # private variable
        self.__age = age # private variable
        self.gender = gender # public variable

def get_name(persn):
    return persn.__name
person = Person("Kartik", 21, "Male")
get_name(person)
```

```python
class Person:
    def __init__(self, name, age, gender):
        self.__name = name # private variable
        self.__age = age # private variable
        self.gender = gender # public variable

def get_name(persn):
    return persn._Person__name # used the dir format (but not accesible by user)
person = Person("Kartik", 21, "Male")
get_name(person)
```

```python
class Person:
    def __init__(self, name, age, gender):
        self._name = name # protected variable
        self._age = age # protected variable
        self.gender = gender

class Employee(Person):
    def __init__(self, name, age, gender):
        super().__init__(name, age, gender)


employee = Employee("Kartik", 21, "Male")
print(employee._name)
```

```python
class Person:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name= name

    def get_age(self):
        return self.__age

    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Age cannot be negative.")

person = Person("Kartik", 21)

print(person.get_name())
print(person.get_age())

person.set_age(22)
print(person.get_age())

person.set_age(-8)
```
